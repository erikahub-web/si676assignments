```python
import requests
import json
from pathlib import Path


baseURL = 'https://www.loc.gov/'
itemEndpoint = 'item/'
searchEndpoint = 'search/'

parameters = {
    'fo' : 'json'
}
```

```python
#1 URI/URL Formulation:
items = [
    "/resource/cph.3f05183/",
    "/resource/fsa.8d24709/",
    "/resource/highsm.64003/"
]

url = [baseURL + item for item in items]
for url in url:
    print(url)
```

```python
#2 LC Permalink URI's:
permalink_baseURL = 'https://lccn.loc.gov/'

lccns = [
    "98508155",
    "2017843202",
    "2020722343"
]

permalinks = [permalink_baseURL + lccn for lccn in lccns]
for permalink in permalinks:
    print(permalink)
```

```python
#3 Retrieve JSON data for items

for lccn in lccns:
    r = requests.get(baseURL + itemEndpoint + lccn, params = parameters)
    
    print(r.url)

    print('response code', r.status_code)
    print('response headers', r.headers)
    
    r.text[:500]

    for element in r.json():
        print(element)

    data = json.loads(r.text)
    for element in data:
        print(element)

fname = 'loc_items'
format = '.json'
n = 0 

for item in lccns:
    n = n + 1
    file = fname + str(n) + format
    
    with open(file, 'w') as f:
        f.write(json.dumps(data, indent = 2))
        print('wrote,file')
print('wrote' ,n, 'files!')
```
